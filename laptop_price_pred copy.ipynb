```python
import re

def safe_extract(text, pattern):
    match = re.search(pattern, str(text))
    return int(match.group(1)) if match else 0

# Handle TB as well by converting to GB
def extract_gb(text, keyword):
    # Match patterns like "1TB HDD", "2TB HDD", "1.0TB HDD"
    match_tb = re.search(r'(\d+\.?\d*)TB\s+' + keyword, str(text))
    # Match patterns like "256GB SSD", "128GB SSD"
    match_gb = re.search(r'(\d+)GB\s+' + keyword, str(text))
    
    if match_tb:
        return int(float(match_tb.group(1)) * 1024)  # Convert TB to GB
    elif match_gb:
        return int(match_gb.group(1))
    return 0

df['SSD_Memory']           = df['Memory'].apply(lambda x: extract_gb(x, 'SSD'))
df['HDD_Memory']           = df['Memory'].apply(lambda x: extract_gb(x, 'HDD'))
df['Flash_Storage_Memory'] = df['Memory'].apply(lambda x: extract_gb(x, 'Flash Storage'))
df['Hybrid_Memory']        = df['Memory'].apply(lambda x: extract_gb(x, 'Hybrid'))

df.head()
```

### Extracting info from 'Cpu' and 'Gpu' column and EDA on it

```python
df.head()
```

```python
df.drop(columns=['ScreenResolution','Memory','Inches','X_res','Y_res'], inplace=True)
```

```python
df.head()
```

```python
df['Cpu'].value_counts()
```

```python
df['Cpu_name'] = df['Cpu'].apply(lambda x: " ".join(x.split()[0:3]))
df.head()
```

```python
df['Cpu_brand'] = df['Cpu_name'].apply(lambda x:
                                       'Intel' if 'Intel' in str(x) else
                                       'AMD' if 'AMD' in str(x) else 'Other')

df['Cpu_tier'] = df['Cpu'].apply(lambda x:
                                 re.search(r'i(\d)', str(x)).group(1)
                                 if re.search(r'i(\d)',str(x)) else 0)

df['Cpu_speed_GHz'] = df['Cpu'].apply(lambda x:
                                      float(re.search(r'(\d+\.\d+)GHz', str(x)).group(1))
                                      if re.search(r'(\d+\.\d+)GHz',str(x)) else 0.0)
df.head()
```

```python
df['Gpu'].value_counts()
```

```python
df['Gpu_brand'] = df['Gpu'].apply(lambda x:
                                  'Nvidia' if 'Nvidia' in str(x) else
                                  'AMD' if 'AMD' in str(x) else
                                  'Intel' if 'Intel' in str(x) else 'Other')

df['Is_dedicated_Gpu'] = df['Gpu'].apply(lambda x:
                                         0 if any(word in str(x) for word in ['UHD', 'HD Graphics', 'Intel', 'Integrated']) else 1)

df.head()
```
