```python
import re

def safe_extract(text, pattern):
    match = re.search(pattern, str(text))
    return int(match.group(1)) if match else 0


def extract_gb(text, keyword):
    
    match_tb = re.search(r'(\d+\.?\d*)TB\s+' + keyword, str(text))
    match_gb = re.search(r'(\d+)GB\s+' + keyword, str(text))
    
    if match_tb:
        return int(float(match_tb.group(1)) * 1024)  # Convert TB to GB
    elif match_gb:
        return int(match_gb.group(1))
    return 0

df['SSD_Memory']           = df['Memory'].apply(lambda x: extract_gb(x, 'SSD'))
df['HDD_Memory']           = df['Memory'].apply(lambda x: extract_gb(x, 'HDD'))
df['Flash_Storage_Memory'] = df['Memory'].apply(lambda x: extract_gb(x, 'Flash Storage'))
df['Hybrid_Memory']        = df['Memory'].apply(lambda x: extract_gb(x, 'Hybrid'))

df.head()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor

cat_columns = ['Company','TypeName','OS','Cpu_brand','Gpu_brand']
num_columns = ['Ram','Weight','SSD_Memory','HDD_Memory','Flash_Storage_Memory','Hybrid_Memory','Touchscreen','IPS_Display','ppi','Cpu_tier','Cpu_speed_GHz','Is_dedicated_Gpu']

cat_pipe = Pipeline(
    [
        ('imputer',SimpleImputer(strategy='most_frequent')),
        ('OHE',OneHotEncoder(handle_unknown='ignore', drop='first')),
        ('scaler',StandardScaler(with_mean=False)),
    ]
)

num_pipe = Pipeline(
    [
        ('imputer',SimpleImputer(strategy='median')),
        ('scaler',StandardScaler())
    ]
)

preprocessor = ColumnTransformer(
    [
        ('col_trnsf',cat_pipe,cat_columns),
        ('num_trnsf',num_pipe,num_columns)
    ], remainder='passthrough'
)

full_pipeline = Pipeline(
    [
        ('preprocessor',preprocessor),
        ('lr',LinearRegression())
    ]
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.metrics import r2_score, mean_absolute_error
from sklearn.svm import SVR                              
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeRegressor          
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import RandomForestRegressor, ExtraTreesRegressor, GradientBoostingRegressor

classifiers = {
    'DTR' : DecisionTreeRegressor(),
    'KNR' : KNeighborsRegressor(),
    'RFR' : RandomForestRegressor(),
    'ETR' : ExtraTreesRegressor(),
    'GBR' : GradientBoostingRegressor()
}

def model_train(clf, X_train, y_train, X_test, y_test):
    full_pipeline = Pipeline([
        ('preprocessor',preprocessor),
        ('model',clf)
    ])
    
    full_pipeline.fit(X_train, y_train)
    y_pred = full_pipeline.predict(X_test)

    r2_score_of_model = r2_score(y_test,y_pred)
    mae_of_model = mean_absolute_error(y_test,y_pred)

    return r2_score_of_model, mae_of_model

r2_scores = []
mae_scores = []

for name, clf in classifiers.items():
    current_r2_scores, current_mae_scores = model_train(clf,
                                        X_train=X_train,
                                        y_train=y_train,
                                        X_test = X_test,
                                        y_test = y_test)
    
    print(name)
    print('R2 score: ',current_r2_scores)
    print('mae score: ',current_mae_scores)

    r2_scores.append(current_r2_scores)
    mae_scores.append(current_mae_scores)
```
